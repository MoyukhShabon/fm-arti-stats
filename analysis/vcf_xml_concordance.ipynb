```python
import glob
import os
import polars as pl
import sys

sys.path.append(os.path.abspath("../common-ffpe-snvf/python"))

from mutation_signatures import read_variants
```

```python
vcf_paths = glob.glob("../data/*.vcf")

all_vcf_snvs = []

for path in vcf_paths:
	vcf_snvs = read_variants(path, sample_name=os.path.basename(path).split(".")[0], snv_only=True, sort_variants=True)
	
	all_vcf_snvs.append(vcf_snvs)

all_vcf_snvs = pl.concat(all_vcf_snvs, how="vertical_relaxed").with_columns(pl.col("pos").cast(int))
all_vcf_snvs
```

```python
xml_snv_paths = glob.glob("../xml-snvs/*/*.snv")

all_xml_snvs = []

for path in xml_snv_paths:
	xml_snvs = read_variants(path, sample_name=os.path.basename(path).split(".")[0], sort_variants=True)
	
	all_xml_snvs.append(xml_snvs)

all_xml_snvs = pl.concat(all_xml_snvs)
all_xml_snvs
```

```python
xml_snv_not_in_vcf = all_xml_snvs.join(all_vcf_snvs, on=["chrom", "pos", "ref", "alt", "sample_id"], how = "anti")
xml_snv_not_in_vcf
```

```python
xml_snv_in_vcf = all_xml_snvs.join(all_vcf_snvs, on=["chrom", "pos", "ref", "alt", "sample_id"], how = "semi")
xml_snv_in_vcf
```

```python
vcf_snv_not_in_xml = all_vcf_snvs.join(all_xml_snvs, on=["chrom", "pos", "ref", "alt", "sample_id"], how = "anti")
vcf_snv_not_in_xml
```

```python
vcf_snv_in_xml = all_vcf_snvs.join(all_xml_snvs, on=["chrom", "pos", "ref", "alt", "sample_id"], how = "semi")
vcf_snv_in_xml
```
